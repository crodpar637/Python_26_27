Partiendo del resultado del experimento anterior, cree un gráfico de barras que muestre en el eje horizontal cada uno de los países y en el eje vertical el número de casos y el número de muertos. Incluya una leyenda para identificar cada uno de los atributos. Guarde esta figura como “covidEuropa.png”

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("https://opendata.ecdc.europa.eu/covid19/casedistribution/csv")
```

```python
df.head()
```

```python
dfEuropa = df[df.continentExp == "Europe"]
```

```python
dfEuropa.head()
```

```python
dfEuropa2 = dfEuropa.groupby(by="countriesAndTerritories")[['cases','deaths']].sum().sort_values('cases',ascending=False)
```

```python
fig, ax = plt.subplots()          # creo la figura y los ejes
axes = dfEuropa2.head().plot.bar(ax=ax)  # pandas dibuja en esos ejes
plt.xticks(rotation=45)            # ahora rota las etiquetas de esa figura
# Añadir título y etiquetas de los ejes
ax.set_title("COVID en Europa")
ax.set_xlabel("Países")
ax.set_ylabel("Número de casos y muertes")

ax.ticklabel_format(style='plain', axis='y') # modificar la representación de números

# Mostrar leyenda (automática en pandas)
ax.legend(title="Atributos");  # 'cases' y 'deaths' aparecerán aquí

#fig.savefig("covidEuropa.png")    # guardo correctamente
```

```python
fig.savefig("covidEuropa.png")
```
